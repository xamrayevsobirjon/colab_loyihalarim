```python
import numpy as np
from numba import cuda
import time

class AITensor:
    def __init__(self, rows, cols):
        self.rows = rows
        self.cols = cols
        self.size_bytes = rows * cols * 4

        self.h_data = cuda.pinned_array((rows, cols), dtype=np.float32)
        self.h_data[:] = np.random.rand(rows, cols).astype(np.float32)

        self.d_data = cuda.device_array((rows, cols), dtype=np.float32)

    def transfer_to_device(self, stream=0):
        self.d_data.copy_to_device(self.h_data, stream=stream)

    def transfer_to_host(self, stream=0):
        self.d_data.copy_to_host(self.h_data, stream=stream)

def initialize_ai_architecture():
    N = 4096
    print(f"[ARXITEKTURA] AI Tenzorlari xotirada yaratilmoqda ({N}x{N})...")

    weight_matrix = AITensor(N, N)
    input_matrix = AITensor(N, N)
    output_matrix = AITensor(N, N)

    print("[ARXITEKTURA] Pinned Memory muvaffaqiyatli ajratildi va initsializatsiya qilindi.")

    stream1 = cuda.stream()
    stream2 = cuda.stream()

    print("[ARXITEKTURA] PCI-e shinasi orqali asinxron uzatish boshlandi...")
    weight_matrix.transfer_to_device(stream=stream1)
    input_matrix.transfer_to_device(stream=stream2)

    stream1.synchronize()
    stream2.synchronize()
    print("[ARXITEKTURA] Ma'lumotlar GPU xotirasiga to'liq yetib bordi va yadro (Kernel) ijrosiga tayyor!")

if __name__ == '__main__':
    initialize_ai_architecture()
```
